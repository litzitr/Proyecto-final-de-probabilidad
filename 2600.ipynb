```python
import glob
import pandas as pd
import matplotlib.pyplot as plt

archivos = glob.glob("2600/*.csv")
limpios = 0

for f in archivos:
    df = pd.read_csv(f, header=4)
    
    # Si el pulso está entre -0.25 y -0.10 V, y no rebota arriba de 0.04 V:
    if (-0.20 < df["Ampl"].min() < -0.12) and (df["Ampl"].max() < 0.04):
        limpios += 1
        if limpios <= 1073: 
            plt.plot(df["Time"], df["Ampl"])

print(f"Total: {len(archivos)} | Limpios: {limpios} | Ruido: {len(archivos) - limpios}")
plt.title("Pulsos Limpios (2600 V)")
plt.xlabel("Tiempo (s)")
plt.ylabel("Amplitud (V)")
plt.show()
```

```python
Pulsos_clean = [limpios / len(archivos)]
print (f"Porcentaje de pulsos limpios: {Pulsos_clean[0] * 100:.2f}%")

Pulsos_ruido = [1 - Pulsos_clean[0]]

print (f"Porcentaje de pulsos de ruido: {Pulsos_ruido[0] * 100:.2f}%")

```
